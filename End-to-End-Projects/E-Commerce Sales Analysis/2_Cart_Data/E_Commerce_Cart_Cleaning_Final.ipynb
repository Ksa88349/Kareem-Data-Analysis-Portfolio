# E-Commerce Cart Data Cleaning

This notebook documents the cleaning and validation process for the `basket_details` dataset.

```
import pandas as pd
import numpy as np

cart = pd.read_csv("cart_details.csv")
```

## Dataset Documentation Note

The dataset description mentions 150,000 basket transactions. However, the downloaded `basket_details.csv` file contains 15,000 records. The analysis in this notebook is based on the downloaded file as provided.

## Data Quality Assessment

```
print("Duplicate rows:", cart.duplicated().sum())
print("\nMissing values:")
print(cart.isnull().sum())
```

### Date Column Assessment

The `basket_date` column contains valid date values and no missing records. The only issue identified was that the column was stored as an object data type.

```
cart["basket_date"] = pd.to_datetime(cart["basket_date"])
cart["basket_date"].dtype
```

```
cart["basket_date"].min(), cart["basket_date"].max()
```

### Basket Count Assessment

The basket_count column was reviewed for missing values and unusual quantities.

```
cart["basket_count"].value_counts().sort_index()
```

```
cart["basket_count"].describe()
```

### Findings

- No missing values were detected.
- No duplicate rows were detected.
- basket_count ranges from 2 to 10.
- No invalid or negative quantities were found.
- basket_date was converted to datetime format.


## Export Clean Dataset

```
cart.to_csv("cart_details_clean.csv", index=False)
```
